```python
forbidden_id = {12, 63, 84}
login_credentials = {12, 1, 5, 8, 3, 8, 84}

intersection_set = forbidden_id.intersection(login_credentials)

if bool(intersection_set):
    print("Intersection set is not empty")
else:
    print("Intersection set is empty")

combined_set = forbidden_id.union(login_credentials)
print(combined_set)
```

```python
contact_list = {
    "John": {"mobile": "555-1234", "work": "555 - 4567", "home": "555 - 7890"},
    "Jim": {"mobile": "555 - 9876", "work": "555 - 7654", "home": "555 - 4321"},
}

while True:
    names = contact_list.keys()

    entry_name = input("Please enter a contact name: ").capitalize()

    if entry_name in names:
        entry_number = input("Which number would you like to call? ").lower()
        print(contact_list.get(entry_name).get(entry_number, "No number found"))
```

```python
contact_list = {
    "John": {"mobile": "555-1234", "work": "555 - 4567", "home": "555 - 7890"},
    "Jim": {"mobile": "555 - 9876", "work": "555 - 7654", "home": "555 - 4321"},
}

while True:
    names = contact_list.keys()

    entry_name = input("Please enter a contact name: ").capitalize()

    if entry_name in names:
        entry_number = input("Which number would you like to call? ").lower()
        print(contact_list.get(entry_name).get(entry_number, "No number found"))
    else:
        print("No such contact")

    quit = input("Would you like to quit? (y/n) ").lower()
    if quit == "y":
        print("Goodbye!")
        break
```
