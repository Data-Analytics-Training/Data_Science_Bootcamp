```python
# Library Program
from time import sleep

from IPython.display import clear_output

books = [
    ["Kürk Mantolu Madonna", "Sabahattin Ali"],
    ["İçimizdeki Şeytan", "Sabahattin Ali"],
    ["Tutunamayanlar", "Oğuz Atay"],
    ["Çalıkuşu", "Reşat Nuri Güntekin"],
]


while True:
    clear_output()

    print("""
    ******************************************************
    **          Welcome to Our Library Program          **
    **                                                  **
    ** Please select the operation you wish to perform  **
    **                                                  **
    **    1. Add New Book                               **
    **    2. Remove Book                                **
    **    3. View Book List                             **
    **    4. Exit                                       **
    **                                                  **
    ******************************************************
  """)

    try:
        task_menu = int(input(""))

        sleep(2)
        clear_output()

        if task_menu == 1:
            book_title = (
                input("Please enter the name of the book you wish to add: ")
                .strip()
                .title()
            )
            book_author = (
                input("Please enter the name of the author of the book: ")
                .strip()
                .title()
            )

            control_book = False

            for book in books:
                if book_title == book[0]:
                    control_book = True

            if control_book == False:
                books.append([book_title, book_author])
                print(f'\n"{book_title}" has been added to the book list.')
            else:
                print("\nThis book is already registered in the library.")


            sleep(2)

        elif task_menu == 2:
            delete_book_author = (
                input("Please enter the author of the book you wish to delete: ")
                .strip()
                .title()
            )

            i2 = 1
            choose_author_books = []

            for book in books:
                if delete_book_author == book[1]:
                    print(f'''{i2}., "{book[0]}", by {book[1]}''')

                    choose_author_books.append([i2, book[0], book[1]])
                    i2 += 1

            try:
                if i2 == 1:
                    print("No books found by this author!")
                else:
                    choose_book = int(input("\nWhich book do you want to delete? "))

                    if choose_book in range(1, i2):
                        chosen_book = choose_author_books[choose_book - 1][1]

                        for book in books:
                            if book[0] == chosen_book:
                                books.remove(book)
                                print(
                                    f'\nYou have checked out the book "{book[0]}", Happy reading.'
                                )

                    else:
                        print("Invalid Selection")

            except ValueError:
                print("Please enter a number!")

            sleep(2)

        elif task_menu == 3:
            print("List of Books Available in the Library:\n")
            i3 = 1
            for book in books:
                print(f'''{i3}. "{book[0]}" by {book[1]}''')
                i3 += 1

            sleep(2)
            skip_list = input("\nPress Enter to continue...")

        elif task_menu == 4:
            print("\nExiting the program...")
            break

        elif task_menu > 4:
            print("Please select a number between 1 and 4.")

    except ValueError:
        print("Please select a number between 1 and 4.")
    except KeyboardInterrupt:
        print("Close by user.")
        break
```
