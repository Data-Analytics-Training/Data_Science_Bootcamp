```python
number_1 = float(input("Please enter first number: "))
number_2 = float(input("Please enter second number: "))

print("Sum of first number and second number is ", number_1 + number_2)
print("Subtraction of first number and second number is ", number_1 - number_2)
print("Multiplication of first number and second number is ", number_1 * number_2)
print("Division of first number and second number is ", number_1 / number_2)
```

```python
number_1 = float(input("Please enter first number: "))
symbol = input("Please enter the operation to be performed: ")
number_2 = float(input("Please enter second number: "))

if symbol == "+":
    print("Sum of first number and second number is ", number_1 + number_2)

if symbol == "-":
    print("Subtraction of first number and second number is ", number_1 - number_2)

if symbol == "*":
    print("Multiplication of first number and second number is ", number_1 * number_2)

if symbol == "/":
    print("Division of first number and second number is ", number_1 / number_2)
```

```python
number_1 = float(input("Please enter first number: "))
symbol = input("Please enter the operation to be performed: ")
number_2 = float(input("Please enter second number: "))

if symbol == "+":
    print("Sum of first number and second number is ", number_1 + number_2)
elif symbol == "-":
    print("Subtraction of first number and second number is ", number_1 - number_2)
elif symbol == "*":
    print("Multiplication of first number and second number is ", number_1 * number_2)
elif symbol == "/":
    print("Division of first number and second number is ", number_1 / number_2)
else:
    print("Invalid symbol")
```

### Account Login Application

```python
user_name = input("Please enter your name: ")
password = input("Please enter your password: ")

defined_username = "johndoe"
defined_password = "Python"

if user_name != defined_username:
    print("Wrong username!")
elif password != defined_password:
    print("Wrong password!")
else:
    print("Welcome!")
```

```python
user_name = input("Please enter your name: ")
password = input("Please enter your password: ")

defined_username = "johndoe"
defined_password = "Python"

if user_name != defined_username or password != defined_password:
    print("Wrong username or password!")
else:
    print("Welcome!")
```

### Student Record Calculation

```python
grade = float(input("Please enter your grade: "))

if grade >= 85:
    print("Congratulations! You Have Earned an Certificate of Excellence")
elif grade >= 70:
    print("You have earned a Certificate of Proficiency")
elif grade < 40:
    print("You failed the course")
else:
    print("You pass the course")
```

```python
grade = float(input("Please enter your grade: "))

if grade >= 85 and grade <= 100:
    print("Congratulations! You Have Earned an Certificate of Excellence")
elif grade >= 70 and grade < 85:
    print("You have earned a Certificate of Proficiency")
elif grade < 70 and grade >= 40:
    print("You have earned a Pass")
elif grade < 40 and grade >= 0:
    print("You failed the course")
else:
    print("You entered an invalid grade")
```

```python
grade = float(input("Please enter your grade: "))

if grade > 100 or grade < 0:
    print("You entered an invalid grade")
elif grade >= 85:
    print("Congratulations! You Have Earned an Certificate of Excellence")
elif grade >= 70:
    print("Congratulations! You Have Earned an  Certificate of Proficiency")
elif grade < 40:
    print("You failed the course")
else:
    print("You have earned a Pass")
```
