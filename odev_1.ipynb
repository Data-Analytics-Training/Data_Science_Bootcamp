```python
from time import sleep
from IPython.display import clear_output

x = 0
print('Aklımdan bir sayı tuttum haydi bil!')
while True:
  y = int(input('Bir sayı giriniz \n(Sayınız pozitif ve tam sayı olmak zorundadır!) '))

  if y < 0:
    print('yanlis giris yaptiniz. Tekrar deneyin.')
    sleep(2)
    clear_output()

  else:
    control = (y*y) - y

    if control == x:
      print('Tebrikler, 100 Puan kazandınız.')
      break
    elif abs(control - x) <= 10:
      print('Tebrikler, 50 Puan kazandınız.')
      break
    elif abs(control - x) <= 50:
      print('Yaklaştın, tekrar deneyin.')
      sleep(3)
      clear_output()
    else:
      print('Kaybettiniz.')

      check = input('Tekrar denemek ister misiniz? (E/H)').lower()
      if check == 'e':
        pass
        clear_output()
      elif check == 'h':
        print('Oyunu bitirdiniz.')
        break
      else:
        print('Yanlış seçim yaptınız, oyundan çıkılıyor.')
        break
```
