### Try - Except - Finally

```python
a = float(input("Please enter dividend number: "))
b = float(input("Please enter divisor number: "))

print(a / b)
```

```python
# ZeroDivisionError: float division by zero
a = float(input("Please enter dividend number: "))  # 10
b = float(input("Please enter divisor number: "))  # 0

print(a / b)
```

```python
# ValueError: could not convert string to float
a = float(input("Please enter dividend number: "))  # john
b = float(input("Please enter divisor number: "))

print(a / b)
```

```python
# try - except example => ValueError
try:
    a = float(input("Please enter dividend number: "))  # john
    b = float(input("Please enter divisor number: "))

    print(a / b)

except ValueError:
    print("Please enter valid numbers")
```

```python
# try - except example => ZeroDivisionError
try:
    a = float(input("Please enter dividend number: "))  # 10
    b = float(input("Please enter divisor number: "))  # 0

    print(a / b)
except ValueError:
    print("Please enter valid numbers")
except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
# try - except example => KeyboardInterrupt
try:
    a = float(input("Enter a number: "))
    b = float(input("Enter another number: "))

    print(a / b)
except ValueError:
    print("Please enter valid numbers")
except ZeroDivisionError:
    print("Cannot divide by zero")
except KeyboardInterrupt:
    print("Program terminated by user")
```

```python
# try - except example => exception handling
try:
    a = float(input("Enter a number: "))
    b = float(input("Enter another number: "))

    print(a / b)

except ValueError:
    print("Please enter valid numbers")
except ZeroDivisionError:
    print("Cannot divide by zero")
except KeyboardInterrupt:
    print("Program terminated by user")
except:
    print("Invalid input")
```

```python
# try - except - finally example =>
try:
    a = float(input("Enter a number: "))
    b = float(input("Enter another number: "))

    print(a / b)

except ValueError:
    print("Please enter valid numbers")
except ZeroDivisionError:
    print("Cannot divide by zero")
except KeyboardInterrupt:
    print("Program terminated by user")
except:
    print("Invalid input")
finally:
    print("Thank you for using this program")
```

### Raise

```python
number = int(input("Enter a number: "))

try:
    if number == 3:
        raise Exception("Dont use 3")
        print("You entered 3")
except Exception as e:
    print(e)
finally:
    print("Goodbye")
```

```python
import numpy as np

data_set = [4, 7, 2, 10, 5, 9, 8, 3]

q1 = np.percentile(data_set, 25)
q2 = np.percentile(data_set, 50)  # median
q3 = np.percentile(data_set, 75)
print("Q1 (first quartile): ", q1)
print("Q2 (second quartile): ", q2)
print("Q3 (third quartile): ", q3)

print("Mean: ", np.mean(data_set))
```
