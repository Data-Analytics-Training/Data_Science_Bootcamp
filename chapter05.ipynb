# Multiple Linear Regression


```python
# importing packages

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
# Importing Dataset
# FOR ONLY GOOGLE COLAB

# from google.colab import drive

# drive.mount("/content/drive")

# data_way = "/content/drive/MyDrive/TechIstanbul/Advertising.csv"

# data = pd.read_csv(data_way)
```

```python
# Importing Dataset
# FOR IDE

data = pd.read_csv("Advertising.csv")
```

```python
X = data[["TV", "Radio", "Newspaper"]]
y = data["Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=34
)
```

```python
# Model Training

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Prediction

y_pred = model.predict(X_test)
```

```python
# Error Calculation

mse = mean_squared_error(y_test, y_pred)
print("Mean Square Error of Regression :", mse)
```

```python
# Coefficient and Intercept Points

print("Coefficients : ", model.coef_)
print("Intercept Point : ", model.intercept_)
```

# Standardization

## Principal Component Regression (PCR)

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
scaler_data = StandardScaler()
X_scaled = scaler_data.fit_transform(X)
```

```python
print(X_scaled)
```

```python
## PCA (Principle Component Analysis)
# Components = 8

pca = PCA(n_components=8)
X_pca = pca.fit_transform(X_scaled)
```

```python
# Model Prediction

y_pred = model.predict(X_test)
```

# Non-Linear Regression

## K-Nearest Neighbors (KNN)

```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsRegressor
```

```python
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data["Target"] = iris.target
```

```python
print(data.head(10))
```

```python
print(data.describe())
```

```python
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# For K = 3
# Model Training

knn = KNeighborsRegressor(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
# Model prediction

y_pred_knn = knn.predict(X_test)
```

```python
# Error Calculation

mse = mean_squared_error(y_test, y_pred_knn)
print("Mean Square Error of KNN Regression (for K = 3) :", mse)
```

```python
# For K = 5
# Model Training

knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X_train, y_train)
```

```python
# Model prediction

y_pred_knn = knn.predict(X_test)
```

```python
# Error Calculation

mse = mean_squared_error(y_test, y_pred_knn)
print("Mean Square Error of KNN Regression (for K = 5) :", mse)
```

```python
# For K = 10
# Model Training

knn = KNeighborsRegressor(n_neighbors=10)
knn.fit(X_train, y_train)
```

```python
# Model prediction

y_pred_knn = knn.predict(X_test)
```

```python
# Error Calculation

mse = mean_squared_error(y_test, y_pred_knn)
print("Mean Square Error of KNN Regression (for K = 10) :", mse)
```

```python
# For K = 25
# Model Training

knn = KNeighborsRegressor(n_neighbors=25)
knn.fit(X_train, y_train)
```

```python
# Model prediction

y_pred_knn = knn.predict(X_test)
```

```python
# Error Calculation

mse = mean_squared_error(y_test, y_pred_knn)
print("Mean Square Error of KNN Regression (for K = 25) :", mse)
```
